# StreamFlix — Phase 1 Data Cleaning & Quality Check

## Objective: 
Validate the quality, consistency, completeness, and integrity of the StreamFlix dataset before performing exploratory data analysis.

## 1. Dataset Overview

The StreamFlix dataset consists of six interconnected CSV files:

- `subscribers.csv`
- `titles.csv`
- `watch_history.csv`
- `ratings.csv`
- `reviews.csv`
- `watchlist.csv`

The dataset contains approximately **979,000 records** covering subscriber and viewing activity over multiple years.

### Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")
```

### Load All CSV Files

```python
import os

DATA_PATH = "../data"

subscribers = pd.read_csv("subscribers.csv")
titles = pd.read_csv("titles.csv")
watch_history = pd.read_csv("watch_history.csv")
ratings = pd.read_csv("ratings.csv")
reviews = pd.read_csv("reviews.csv")
watchlist = pd.read_csv("watchlist.csv")
```

```python
print("Subscribers:", subscribers.shape)
print("Titles:", titles.shape)
print("Watch History:", watch_history.shape)
print("Ratings:", ratings.shape)
print("Reviews:", reviews.shape)
print("Watchlist:", watchlist.shape)
```

### Create a Dataset Dictionary

```python
datasets = {
    "subscribers": subscribers,
    "titles": titles,
    "watch_history": watch_history,
    "ratings": ratings,
    "reviews": reviews,
    "watchlist": watchlist
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
profile = []

for name, df in datasets.items():
    profile.append({
        "Table": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1]
    })

profile_df = pd.DataFrame(profile)

display(profile_df)
```

### Check Data Types

```python
for name, df in datasets.items():
    print("=" * 80)
    print(f"{name.upper()} DATA TYPES")
    print(df.dtypes)
```

### Create a Complete Data-Type Report

```python
dtype_report = []

for name, df in datasets.items():
    for column in df.columns:
        dtype_report.append({
            "Table": name,
            "Column": column,
            "Data Type": str(df[column].dtype)
        })

dtype_report_df = pd.DataFrame(dtype_report)

display(dtype_report_df)
```

### Check Missing Values

```python
def missing_value_report(df, table_name):
    report = pd.DataFrame({
        "Table": table_name,
        "Column": df.columns,
        "Missing_Count": df.isnull().sum().values,
        "Missing_Percentage": (
            df.isnull().sum().values / len(df) * 100
        )
    })
    
    return report

missing_reports = []

for name, df in datasets.items():
    missing_reports.append(
        missing_value_report(df, name)
    )

missing_df = pd.concat(missing_reports, ignore_index=True)

display(missing_df)
```

### Show Only Columns Having Missing Values

```python
missing_only = missing_df[
    missing_df["Missing_Count"] > 0
].sort_values(
    ["Table", "Missing_Percentage"],
    ascending=[True, False]
)

display(missing_only)
```

### Check Duplicate Rows

```python
duplicate_report = []

for name, df in datasets.items():
    duplicate_report.append({
        "Table": name,
        "Duplicate_Rows": df.duplicated().sum(),
        "Duplicate_Percentage": (
            df.duplicated().sum() / len(df) * 100
        )
    })

duplicate_df = pd.DataFrame(duplicate_report)

display(duplicate_df)
```

### Check Duplicate Subscriber IDs

```python
subscriber_id_duplicates = subscribers[
    subscribers["subscriber_id"].duplicated(keep=False)
]

print(
    "Duplicate subscriber_id records:",
    len(subscriber_id_duplicates)
)

display(subscriber_id_duplicates.head(20))
```

### Check Duplicate Watch IDs

```python
watch_id_duplicates = watch_history[
    watch_history["watch_id"].duplicated(keep=False)
]

print(
    "Duplicate watch_id records:",
    len(watch_id_duplicates)
)

display(watch_id_duplicates.head(20))
```

### Convert Date Columns

```python
date_columns = {
    "watch_history": ["watch_date"],
    "subscribers": ["signup_date", "churn_date"]
}

for table_name, columns in date_columns.items():
    df = datasets[table_name]
    
    for column in columns:
        if column in df.columns:
            df[column] = pd.to_datetime(
                df[column],
                errors="coerce"
            )
```

### Verify Date Conversion

```python
print("Watch Date:")
print(watch_history["watch_date"].dtype)

print("\nSubscriber Signup Date:")
print(subscribers["signup_date"].dtype)

print("\nSubscriber Churn Date:")
print(subscribers["churn_date"].dtype)
```

### Convert Important Numeric Columns

```python
numeric_columns = {
    "watch_history": [
        "watch_duration_min",
        "completion_pct"
    ],
    "subscribers": [
        "monthly_price_usd",
        "age",
        "tenure_months"
    ],
    "titles": [
        "license_cost_usd",
        "total_watch_hours",
        "total_plays"
    ]
}

for table_name, columns in numeric_columns.items():
    df = datasets[table_name]
    
    for column in columns:
        if column in df.columns:
            df[column] = pd.to_numeric(
                df[column],
                errors="coerce"
            )
```

### Detect Negative Watch Duration

```python
negative_duration = watch_history[
    watch_history["watch_duration_min"] < 0
]

print(
    "Negative watch duration records:",
    len(negative_duration)
)

display(negative_duration.head())
```

### Check Completion Percentage

```python
invalid_completion = watch_history[
    (watch_history["completion_pct"] < 0) |
    (watch_history["completion_pct"] > 100)
]

print(
    "Invalid completion percentage records:",
    len(invalid_completion)
)

display(invalid_completion.head())
```

### Check Outliers in Watch Duration

```python
watch_history["watch_duration_min"].describe(
    percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
)
```

```python
Q1 = watch_history["watch_duration_min"].quantile(0.25)
Q3 = watch_history["watch_duration_min"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

duration_outliers = watch_history[
    (watch_history["watch_duration_min"] < lower_bound) |
    (watch_history["watch_duration_min"] > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print(
    "\nPotential statistical outliers:",
    len(duration_outliers)
)
```

### Check Referential Integrity: subscriber_id

```python
valid_subscribers = set(
    subscribers["subscriber_id"].dropna()
)

invalid_watch_subscribers = watch_history[
    ~watch_history["subscriber_id"].isin(valid_subscribers)
]

print(
    "Watch history records with invalid subscriber_id:",
    len(invalid_watch_subscribers)
)

display(
    invalid_watch_subscribers.head(20)
)
```

### Check Referential Integrity: title_id

```python
valid_titles = set(
    titles["title_id"].dropna()
)

invalid_watch_titles = watch_history[
    ~watch_history["title_id"].isin(valid_titles)
]

print(
    "Watch history records with invalid title_id:",
    len(invalid_watch_titles)
)

display(
    invalid_watch_titles.head(20)
)
```

### Check Subscribers: Churn Date vs Signup Date

```python
invalid_churn_dates = subscribers[
    subscribers["churn_date"].notna() &
    subscribers["signup_date"].notna() &
    (subscribers["churn_date"] <= subscribers["signup_date"])
]

print(
    "Invalid churn date records:",
    len(invalid_churn_dates)
)

display(invalid_churn_dates.head(20))
```

### Check Active Subscribers

```python
active_with_churn_date = subscribers[
    (subscribers["is_active"] == True) &
    (subscribers["churn_date"].notna())
]

print(
    "Active subscribers with churn_date:",
    len(active_with_churn_date)
)

display(active_with_churn_date.head(20))
```

### Check Churned Subscribers

```python
churned_without_date = subscribers[
    (subscribers["is_active"] == False) &
    (subscribers["churn_date"].isna())
]

print(
    "Inactive subscribers without churn_date:",
    len(churned_without_date)
)

display(churned_without_date.head(20))
```

### Validate Completion Percentage Against Duration

```python
print("Watch History columns:")
print(watch_history.columns.tolist())

print("\nTitles columns:")
print(titles.columns.tolist())
```

```python
"content_duration_min" in watch_history.columns
```

```python
"content_duration_min" in titles.columns
```

```python
watch_history["calculated_completion_pct"] = (
    watch_history["watch_duration_min"] /
    watch_history["content_duration_min"] * 100
)

watch_history["completion_difference"] = (
    watch_history["completion_pct"] -
    watch_history["calculated_completion_pct"]
)

completion_mismatch = watch_history[
    watch_history["completion_difference"].abs() > 5
]

print(
    "Potential completion mismatches:",
    len(completion_mismatch)
)
```
